```python
import pyodbc
import pandas as pd
```

```python
server = r"LAPTOP-PUH0OFS5"
database = "pick_n_pay"

connection_string = (
    "DRIVER={ODBC Driver 18 for SQL Server};"
    f"SERVER={server};"
    f"DATABASE={database};"
    "Trusted_Connection=yes;"
    "TrustServerCertificate=yes;"
)
```

```python
conn = pyodbc.connect(connection_string)

print("Successfully connected to SQL Server!")
```

```python
query = """
SELECT top 2
    [transaction_id],
    [timestamp],
    [store_location],
    [sku],
    [product_name],
    [category],
    [quantity],
    [unit_price_zar],
    [total_amount_zar],
    [payment_method]
FROM [dbo].[combined_beverage_sales]
"""
```

```python
top1 = """
SELECT top 1
    [transaction_id],
    [timestamp],
    [store_location],
    [sku],
    [product_name],
    [category],
    [quantity],
    [unit_price_zar],
    [total_amount_zar],
    [payment_method]
FROM [dbo].[combined_beverage_sales]
"""
```

```python
df1 = pd.read_sql(top1, conn)

print("Data successfully loaded into DataFrame!")
```

```python
df1.head()
```

```python
df = pd.read_sql(query, conn)

print("Data successfully loaded into DataFrame!")
```
